```python
from pathlib import Path
import pandas as pd
import numpy as np

SHOPEE_DIR = Path(
    "/kaggle/input/datasets/trnquangtriu/10dataset/Shopee Rebrowser"
)

parquet_files = sorted(SHOPEE_DIR.glob("*.parquet"))

if not parquet_files:
    raise FileNotFoundError(f"Không có Parquet tại {SHOPEE_DIR}")

frames = []

for path in parquet_files:
    df = pd.read_parquet(path)
    df["source_file"] = path.name
    frames.append(df)

shopee_raw = pd.concat(frames, ignore_index=True)

print("SỐ FILE:", len(parquet_files))
print("TỔNG BẢN GHI:", len(shopee_raw))
print("SỐ CỘT:", len(shopee_raw.columns))

# Chuẩn hóa giá trị placeholder
PLACEHOLDERS = {
    "",
    "[PREMIUM]",
    "NONE",
    "NULL",
    "NAN",
    "N/A"
}

def usable(series):
    s = series.astype("string").str.strip()
    return (
        s.notna()
        & ~s.str.upper().isin(PLACEHOLDERS)
    )

print("\n1. PHÂN BỐ ĐƠN VỊ TIỀN TỆ")
display(
    shopee_raw["currency"]
    .fillna("MISSING")
    .value_counts(dropna=False)
    .rename_axis("currency")
    .reset_index(name="rows")
)

print("\n2. SỐ BẢN GHI THEO FILE")
display(
    shopee_raw["source_file"]
    .value_counts()
    .sort_index()
    .rename_axis("file")
    .reset_index(name="rows")
)

print("\n3. KIỂM TRA TRÙNG SẢN PHẨM")

key_cols = ["shopId", "itemId"]

valid_key = shopee_raw[key_cols].notna().all(axis=1)

print("Bản ghi có đủ shopId + itemId:", valid_key.sum())

print(
    "Sản phẩm duy nhất:",
    shopee_raw.loc[valid_key, key_cols]
    .drop_duplicates()
    .shape[0]
)

print(
    "Bản ghi trùng shopId + itemId:",
    shopee_raw.loc[valid_key]
    .duplicated(key_cols)
    .sum()
)

print("\n4. CHẤT LƯỢNG CÁC TRƯỜNG")

quality_cols = [
    "title",
    "description",
    "brand",
    "currency",
    "price",
    "priceMin",
    "priceMax",
    "ratingAvg",
    "ratingCount",
    "commentCount",
    "productUrl",
    "shopLocation"
]

quality = []

for col in quality_cols:
    if col not in shopee_raw.columns:
        continue

    count = int(usable(shopee_raw[col]).sum())

    quality.append({
        "column": col,
        "usable_rows": count,
        "usable_pct": round(
            count / len(shopee_raw) * 100, 2
        )
    })

display(pd.DataFrame(quality))

print("\n5. KIỂM TRA URL THẬT")

url_text = shopee_raw["productUrl"].astype("string").str.strip()

valid_url = url_text.str.match(
    r"^https?://",
    case=False,
    na=False
)

print("URL có dạng http/https:", valid_url.sum())

display(
    shopee_raw.loc[
        valid_url,
        ["itemId", "shopId", "currency", "productUrl"]
    ].head(10)
)

print("\n6. KIỂM TRA PRICE MIN/MAX")

for col in ["priceMin", "priceMax"]:
    values = pd.to_numeric(
        shopee_raw[col],
        errors="coerce"
    )

    print(f"\n{col}")
    print(values.describe())

print("\n7. MẪU SẢN PHẨM THEO CURRENCY")

sample_cols = [
    "itemId",
    "shopId",
    "title",
    "currency",
    "price",
    "priceMin",
    "priceMax",
    "ratingAvg",
    "productUrl"
]

for currency, group in shopee_raw.groupby(
    "currency",
    dropna=False
):
    print("\nCURRENCY:", currency)
    display(group[sample_cols].head(3))
```

```python
import requests
import json
import pandas as pd
from pathlib import Path
from tqdm.auto import tqdm

OUT_DIR = Path("/kaggle/working/step10")
OUT_DIR.mkdir(parents=True, exist_ok=True)

def normalize_text(value):
    if value is None:
        return ""

    if isinstance(value, list):
        return "\n".join(
            str(x).strip()
            for x in value
            if x is not None and str(x).strip()
        )

    return str(value).strip()


matched_rows = []
matched_asins = set()
scanned = 0

with requests.get(
    META_URL,
    stream=True,
    timeout=(30, 180)
) as response:

    response.raise_for_status()

    for line in tqdm(
        response.iter_lines(),
        desc="Scanning Electronics metadata"
    ):

        if not line:
            continue

        item = json.loads(line)
        scanned += 1

        asin = str(
            item.get("parent_asin") or ""
        ).strip().upper()

        if asin not in target_asins:
            continue

        matched_asins.add(asin)

        matched_rows.append({
            "asin": asin,
            "meta_title": normalize_text(
                item.get("title")
            ),
            "meta_description": normalize_text(
                item.get("description")
            ),
            "meta_features": normalize_text(
                item.get("features")
            ),
            "meta_categories": normalize_text(
                item.get("categories")
            ),
            "meta_main_category": normalize_text(
                item.get("main_category")
            ),
            "meta_store": normalize_text(
                item.get("store")
            ),
            "metadata_source": "Amazon Reviews 2023",
            "metadata_category": "Electronics"
        })

metadata_electronics = pd.DataFrame(matched_rows)

# Chỉ giữ một metadata cho mỗi ASIN.
# Nếu có nhiều bản ghi, ưu tiên bản có nội dung đầy đủ hơn.
if not metadata_electronics.empty:

    text_cols = [
        "meta_description",
        "meta_features",
        "meta_categories"
    ]

    metadata_electronics["content_score"] = sum(
        metadata_electronics[c].str.len().fillna(0)
        for c in text_cols
    )

    metadata_electronics = (
        metadata_electronics
        .sort_values(
            "content_score",
            ascending=False
        )
        .drop_duplicates(
            "asin",
            keep="first"
        )
        .drop(
            columns="content_score"
        )
        .reset_index(drop=True)
    )

metadata_electronics.to_parquet(
    OUT_DIR / "amazon_metadata_electronics_matched.parquet",
    index=False
)

print("\n===== KẾT QUẢ =====")
print("Metadata đã quét:", scanned)
print("ASIN khớp duy nhất:", len(metadata_electronics))

print(
    "Tỷ lệ bao phủ Amazon Step 9:",
    round(
        len(metadata_electronics)
        / len(target_asins) * 100,
        2
    ),
    "%"
)

if not metadata_electronics.empty:

    for col in [
        "meta_description",
        "meta_features",
        "meta_categories"
    ]:
        print(
            col,
            "có dữ liệu:",
            metadata_electronics[col]
            .str.strip()
            .ne("")
            .sum()
        )

    display(metadata_electronics.head(5))
```

```python
from pathlib import Path
import pandas as pd

OUT_DIR = Path("/kaggle/working/step10")
OUT_DIR.mkdir(parents=True, exist_ok=True)

PRODUCTS_PATH = Path(
    "/kaggle/input/datasets/trnquangtriu/"
    "dataset10b/smart_products.parquet"
)

REVIEWS_PATH = Path(
    "/kaggle/input/datasets/trnquangtriu/"
    "dataset10c/smart_reviews.parquet"
)

METADATA_PATH = (
    OUT_DIR / "amazon_metadata_electronics_matched.parquet"
)

# Kiểm tra file tồn tại
for path in [
    PRODUCTS_PATH,
    REVIEWS_PATH,
    METADATA_PATH
]:
    if not path.exists():
        raise FileNotFoundError(f"Không tìm thấy: {path}")

# Đọc dữ liệu
products = pd.read_parquet(PRODUCTS_PATH)
reviews = pd.read_parquet(REVIEWS_PATH)
metadata = pd.read_parquet(METADATA_PATH)

print("===== DỮ LIỆU ĐẦU VÀO =====")
print("Step 9 products:", products.shape)
print("Step 9 reviews:", reviews.shape)
print("Amazon metadata:", metadata.shape)

# Kiểm tra số lượng
assert len(products) == 164_926
assert len(reviews) == 767_179
assert len(metadata) == 121_917

# Kiểm tra khóa
assert products["product_id"].is_unique
assert metadata["asin"].is_unique

print("\nĐã đọc dữ liệu thành công!")
```

```python
import pandas as pd

unified_products = products.copy()

unified_products["asin"] = pd.NA

amazon_mask = (
    unified_products["source"]
    .astype("string")
    .str.lower()
    .eq("amazon")
    .fillna(False)
)

unified_products.loc[amazon_mask, "asin"] = (
    unified_products.loc[amazon_mask, "product_id"]
    .astype("string")
    .str.replace(r"^amazon_", "", regex=True)
    .str.upper()
    .str.strip()
)

metadata_cols = [
    "asin",
    "meta_title",
    "meta_description",
    "meta_features",
    "meta_categories",
    "meta_main_category",
    "meta_store",
    "metadata_source",
    "metadata_category"
]

unified_products = unified_products.merge(
    metadata[metadata_cols],
    on="asin",
    how="left",
    validate="many_to_one"
)

# Chỉ điền description khi bản gốc thiếu
original_description = (
    unified_products["description"]
    .astype("string")
    .str.strip()
)

missing_description = (
    original_description.isna()
    | original_description.eq("")
    | original_description.str.lower().isin(
        ["none", "nan", "null"]
    )
).fillna(True)

new_description = (
    unified_products["meta_description"]
    .astype("string")
    .str.strip()
)

valid_new_description = (
    new_description.notna()
    & new_description.ne("")
).fillna(False)

fill_mask = missing_description & valid_new_description

unified_products.loc[
    fill_mask,
    "description"
] = unified_products.loc[
    fill_mask,
    "meta_description"
]

# Cột bổ sung phục vụ tìm kiếm và RAG
unified_products["product_features"] = (
    unified_products["meta_features"]
)

unified_products["category_path"] = (
    unified_products["meta_categories"]
)

unified_products["metadata_enriched"] = (
    unified_products["metadata_source"].notna()
)

unified_products["description_enriched"] = fill_mask

print("Tổng sản phẩm:", len(unified_products))
print(
    "Sản phẩm khớp metadata:",
    unified_products["metadata_enriched"].sum()
)
print(
    "Description được bổ sung:",
    unified_products["description_enriched"].sum()
)

assert len(unified_products) == len(products)

display(
    unified_products.loc[
        unified_products["description_enriched"],
        [
            "product_id",
            "product_name",
            "description",
            "product_features",
            "category_path"
        ]
    ].head(5)
)
```
